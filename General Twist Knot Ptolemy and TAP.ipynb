```python
R=PolynomialRing(QQ, x)
R.inject_variables()

def pol(n):
    twist = n
    r={}
    if twist >= 3:
        r[4]=(-x^2-(-1)^(twist))/(x)
    if twist >= 4:
        r[6]=(x^4+(-1)^(twist)*x^2+1)/(x^3)
    if twist >= 5:
        i = 4
        while 2*i <= 2*twist - 2:
            r[2*i]=(r[2*i-2]^2+(-1)^(twist-i))/(r[2*i-4])
            i = i + 1

    partial=r[2*twist-2]^2-r[2*twist-4]
    p=partial.numerator()+partial.denominator()

    return(p/((-1)^(twist+1)*r[2*twist-4].numerator()))

def populate_poly(n=50):
    poly={}
    poly[2]=x^2+x+1
    poly[3]=x^3-x-1

    for i in range(4,n+1):
        poly[i]=pol(i)/poly[i-2]
    return poly
        
def tap(n): 
    twist=n
    
    poly=populate_poly(twist)
        
    S.<w> = NumberField(R('{}'.format(poly[twist])))
    T.<t> = LaurentPolynomialRing(S)

    M=MatrixSpace(T,2)

    a=t*M.matrix([1, 1/w^2,
                0, 1])
    b=t*M.matrix([1, 0,
                -1/w^2, 1])
    A=a.inverse()
    B=b.inverse()

    Id=M.identity_matrix()

    if twist%2 == 0:
        m = twist // 2
        word = (A*b)^m*(a*B)^m
        rel = word*a-b*word
        fox = M.zero_matrix()
        for i in range(m):
            fox = fox - (A*b)^(i)*A
        for i in range(m+1):
            fox = fox + (A*b)^m*(a*B)^i
        for i in range(m):
            fox = fox + (b*A)^(i+1)
        for i in range(m):
            fox = fox - (b*A)^m*b*(a*B)^i

    else:
        m = (twist // 2) + 1
        word = (A*b)^m*(a*B)^m
        rel = a*word-word*b
        fox = M.zero_matrix()
        for i in range(m):
            fox = fox - (A*b)^(i)*A
        for i in range(m):
            fox = fox + (A*b)^m*(a*B)^i
        for i in range(m-1):
            fox = fox + (b*A)^(i+1)
        for i in range(m):
            fox = fox - (b*A)^(m-1)*b*(a*B)^i

    PDen = Id - b        

    Num = fox.det()
    Den = PDen.det()

    print(Num/Den)
    print(S)
    
```
