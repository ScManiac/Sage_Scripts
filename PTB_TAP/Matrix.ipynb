```python
#Sage doesn't want to convert the 2x2 arrays we get into matrices, so we write our own functions.
#This is only written with 2x2 matrices in mind.

class MyMat():
    def __init__(self, array):
        self.array = array
        
    def __getitem__(self, key):
        return self.array[key]
    
    def __setitem__(self, key, item):
        self.array[key] = item
        
    def __repr__(self):
        end0 = max(len(f'{self[0][0]}'), len(f'{self[1][0]}')) + 3
        end1 = max(len(f'{self[0][1]}'), len(f'{self[1][1]}')) + 3
        
        m00=f'{self[0][0]}'
        m01=f'{self[0][1]}'
        m10=f'{self[1][0]}'
        m11=f'{self[1][1]}'
                      
        return f'{m00 :<{end0}}{m01 :<{end1}}\n{m10 :<{end0}}{m11 :<{end1}}'
    
    def __add__(self, other):
        sum_mat = MyMat([[0,0],[0,0]])
        for row in range(2):
            for col in range(2):
                sum_mat[row][col] = self[row][col] + other[row][col]
        
        return sum_mat
    
    def __rmul__(self, other):
        mult_mat = MyMat([[0,0],[0,0]])
        
        for row in range(2):
            for col in range(2):
                mult_mat[row][col] = self[row][col]*other
        
        return mult_mat
    
    def __mul__(self, other):
        mult_mat = MyMat([[0,0],[0,0]])
        
        if type(other) == type(self):
            mult_mat[0][0] = self[0][0] * other[0][0] + self[0][1] * other[1][0]
            mult_mat[0][1] = self[0][0] * other[0][1] + self[0][1] * other[1][1]
            mult_mat[1][0] = self[1][0] * other[0][0] + self[1][1] * other[1][0]
            mult_mat[1][1] = self[1][0] * other[0][1] + self[1][1] * other[1][1]
                
            return mult_mat
        
        else:
            return self.__rmul__(other)
    
    
    
    def __sub__(self, other):
        return self + (-1)*other
    
    def __pow__(self, other):
        pow_mat = MyMat([[1,0],[0,1]])
        for i in range(other):
            pow_mat = pow_mat*self
        return pow_mat
    
    def __neg__(self):
        return (-1)*self
        
    

                
```
