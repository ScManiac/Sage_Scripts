```python
%run pi1.ipynb
%run monodromyfunctions.ipynb
%run ptolemy_equations.ipynb\

def apply_dict_to_str(dictionary:dict, word:str):
    for key in dictionary.keys():
        word = word.replace(key, dictionary[key])
    return word


#This class will compute all boundary parabolic representations of the given PTB and tensor them with the abelianization homomorphism.
class Representations():
    def __init__(self, monodromy:str):
        self.monodromy = monodromy
        self.pi1 = Pi1(monodromy)
        self.h1 = self.pi1.homology
        
        self.ptolemy = PtolemyEquations(self.monodromy)
        
        temp_gluing_dict = self.ptolemy.gluing_dict
        temp_gluing_dict['t_0'] = '1'
        self.gluing_dict = temp_gluing_dict
        
        self.ring_trivial = self.ptolemy.sage_trivial_ptolemy.coordinate_ring()
        self.ring_monodromy = self.ptolemy.sage_monodromy_ptolemy.coordinate_ring()
        self.ring_meridean = self.ptolemy.sage_meridean_ptolemy.coordinate_ring()
        self.ring_longitude = self.ptolemy.sage_longitude_ptolemy.coordinate_ring()
        self.ring_mon_mer = self.ptolemy.sage_mon_mer_ptolemy.coordinate_ring()
        self.ring_mon_long = self.ptolemy.sage_mon_long_ptolemy.coordinate_ring()
        self.ring_mer_long = self.ptolemy.sage_mer_long_ptolemy.coordinate_ring()
        self.ring_all = self.ptolemy.sage_all_ptolemy.coordinate_ring()
        
        self.mat_trivial = self.make_matrix_space(self.ring_trivial)
        self.mat_monodromy = self.make_matrix_space(self.ring_monodromy)
        self.mat_meridean = self.make_matrix_space(self.ring_meridean)
        self.mat_longitude = self.make_matrix_space(self.ring_longitude)
        self.mat_mon_mer = self.make_matrix_space(self.ring_mon_mer)
        self.mat_mon_long = self.make_matrix_space(self.ring_mon_long)
        self.mat_mer_long = self.make_matrix_space(self.ring_mer_long)
        self.mat_all = self.make_matrix_space(self.ring_all)
        
        
    def make_group_ring(self, ring):

        return self.h1.algebra(ring)
    
    def make_matrix_space(self, ring):
        
        return MatrixSpace(ring, 2, 2)
    
    def repeat_gluing_dict(self, word:str):
        end = len(self.monodromy)
    
        for w in range(end):
            word = apply_dict_to_str(self.gluing_dict, word)
        
        return word
    
    def replace_in_matrix(self, matrix):
        end = len(self.monodromy) - 1

        for i in range(2):
            for j in range(2):
                matrix[i][j] = self.repeat_gluing_dict(matrix[i][j])
            
        return matrix
    
    def make_generic_mer_representation(self):
        end = len(self.monodromy) - 1

        return self.replace_in_matrix([[f'(1 + v_0^2 - t_{end}^2)/(v_0)', f'-1/t_{end}'], [f't_{end}', '0']])
    
    def make_generic_long_representation(self):
        end = len(self.monodromy) - 1
        
        return self.replace_in_matrix([[f'(1/t_{end}', f'-v_0/(t_{end})^2'], [f'v_0', f'(t_{end}^2 - v_0^2)/t_{end}']])
        
        
        
```
