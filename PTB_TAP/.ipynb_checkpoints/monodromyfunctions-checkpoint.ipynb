```python
#monodromyfunctions.py>


import numpy as np
from math import gcd

rmatrix = matrix([[1,1],[0,1]])
lmatrix = matrix([[1,0],[1,1]])

rldict = {'R':rmatrix, 'L':lmatrix}
#This function will take a monodromy RL word and translate it into an associated matrix.


def wordtomatrix(word:str):
    translated = matrix.identity(2)
    
    for char in word:
        translated = translated * rldict[char]
    
    return translated


#The classical euclidean division algorithm will be useful throughout including in order to reduce a matrix into an RL word.
#Recall that this will take two numbers m and n, and output two new values q and r where r<n and m = nq + r.

def eucliddiv(num:int, divisor:int):
    quotient = 0
    remainder = num
    sgn = np.sign(num*divisor)
    
    while remainder >= abs(divisor) or remainder < 0:
        quotient += sgn * 1
        remainder = remainder - sgn*divisor
    
    return (quotient, remainder,)

#Sometimes an RL word will be in a form which does not begin with R or doesn't end with L. In this case, we should cycle the letters of the word until this is the case.

def cycleback(word:str):
    return word[1:]+ word[0]

def cycleforward(word:str):
    return word[-1] + word[:-1]

def standardword(word:str):
    
    #The loop will run forever if the word is just a repetition of either R or L, so we will run an initial check just to be safe.
    if 'R' not in word or 'L' not in word:
        return word
    
    while word[0] != 'R':
        word = cycleback(word)
    
    while word[-1] != 'L':
        word = cycleforward(word)
        
    return word

#When we would like to cycle through the vairous valid presentations of the monodromy word, we can use this function.

def differentword(word:str, ver = 0):
    
    word = standardword(word)
    
    for i in range(ver):
        
        while word[0] == 'R':
            word = cycleback(word)
            
        word = standardword(word)
    
    return word

#Row opperations for a 2x2 matrix.

def add0to1(mat:sage.matrix, mult):
    mat[1,0] = mat[1,0] + mult*mat[0,0]
    mat[1,1] = mat[1,1] + mult*mat[0,1]
    
    return mat
    
def add1to0(mat:sage.matrix, mult):
    mat[0,0] = mat[0,0] + mult*mat[1,0]
    mat[0,1] = mat[0,1] + mult*mat[1,1]
    
    return mat


#This function will take a monodromy matrix and translate it to an RL string.
#This involves the Gaussian elimintation algorithm in order to find a sequence of type 3 elementary moves which will reduce the matrix to the identity. 
#Since each type 3 move can be exhibited via an R or L matrix, this results in an RL sequence if the reduction is possible.

def matrixtoword(mat:sage.matrix):
    word = ''
    alt = 1
    trans_dict = {48:'L', 49:'R'}
    
    #First ensure that it is even possible to reduce the first column to (1,0).  
    
    if math.gcd(mat[0,0], mat[1,0]) != 1:
        raise ValueError('No RL decomposition exists for this matrix')
        
    #Reduce the rows of the matrix by alternating 'add row 0 to row 1' and 'add row 1 to row 0'
    #Multiples are determined by the Euclidean algorithm, and the steps taken are recorded in 'word'
    
    while mat[0,0] != 1 and mat[1,0] != 1:
        euc = eucliddiv(mat[(alt + 1)%2, 0], mat[alt , 0])
        
        mat[(alt + 1)%2, 0] = euc[1]
        mat[(alt + 1)%2, 1] = mat[(alt + 1)%2, 1] - euc[0]*mat[alt , 1]
        
        word = word + f'{alt}'* euc[0] 
        
        alt = (alt + 1)%2
        
    #If there is a 1 in the position (1,0), use it to make a 1 in the position (0,0)
    
    if mat[1,0] == 1:
        word = word + '1'*(mat[0,0] - 1)
        mat = add1to0(mat, -(mat[0,0] - 1))
        
    #Use the 1 in the position (0,0) to eliminate the number in the position (1,0)
        
    word = word + '0'*mat[1,0]
    mat = add0to1(mat, -mat[1,0]) 
    
    #Check that it is possible to reduce to the identity, i.e. is there a 1 in the (1,1) position?
    
    if mat[1,1] != 1:
        raise ValueError('No RL decomposition exists for this matrix')
        
    #Reduce to the identity
        
    word = word + '1'*(mat[0,1])
    mat = add1to0(mat, -(mat[0,1]))
    
    return word.translate(trans_dict)
    
```
