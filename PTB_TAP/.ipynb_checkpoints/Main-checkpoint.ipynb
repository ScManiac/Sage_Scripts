```python
%run pi1.ipynb
%run monodromyfunctions.ipynb
%run ptolemy_equations.ipynb\
%run Matrix.ipynb
%run representation.ipynb

monodromy = 'RL'

m = Representations(monodromy).make_generic_mer_representation()
l = Representations(monodromy).make_generic_long_representation()
t = Representations(monodromy).make_generic_monodromy_representation()

rep_dict = {'1':identity_matrix(2), 't' : t, 'm' : m, 'l' : l}

M = [[Pi1(monodromy).fox_derivative_rel1('m'), Pi1(monodromy).fox_derivative_rel1('l')],[Pi1(monodromy).fox_derivative_rel2('m'), Pi1(monodromy).fox_derivative_rel2('l')]]
        
# m00 = -t^-1 + (rep_dict[l] for w in M[0][0].split('+') for 

m00 = matrix([[0,0],[0,0]])
skip = True
for w in M[0][0].split(' + '):
    q = 1
    for e in w.split('*'):
        if not skip:
            q = q * rep_dict[e]
        else:
            q = -t^-1
            skip = False
       
    m00 = m00 + q

m01 = matrix([[0,0],[0,0]])
for w in M[0][1].split(' + '):
    q = 1
    for e in w.split('*'):
        q = q * rep_dict[e]
       
    m01 = m01 + q

m10 = matrix([[0,0],[0,0]])
for w in M[1][0].split(' + '):
    q = 1
    for e in w.split('*'):
        q = q * rep_dict[e]
       
    m10 = m10 + q
    
m11 = matrix([[0,0],[0,0]])
skip = True
for w in M[1][1].split(' + '):
    q = 1
    for e in w.split('*'):
        if not skip:
            q = q * rep_dict[e]
        else:
            q = -t^-1
            skip = False
       
    m11 = m11 + q
    
big_mat = matrix([[m00[0][0], m00[0][1], m01[0][0], m01[0][1]], [m00[1][0], m00[1][1], m01[1][0], m01[1][1]], [m10[0][0], m10[0][1], m11[0][0], m11[0][1]], [m10[1][0], m10[1][1], m11[1][0], m11[1][1]]])
```
