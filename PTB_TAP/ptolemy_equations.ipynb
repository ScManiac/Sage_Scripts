```python
#A function used to merge two dictionaries.
def merge(dict1, dict2):
    res = {**dict1, **dict2}
    return res

class PtolemyEquations():
    def __init__(self, monodromy):
        self.monodromy = monodromy
        self.ptolemy_equations = self.make_ptolemy_equations(monodromy)
        self.trivial_obstruction = self.list_obstructions()
        self.monodromy_obstruction = self.make_monodromy_obstruction_class()
        self.meridean_obstruction = self.make_meridean_obstruction_class()
        self.longitude_obstruction = self.make_longitude_obstruction_class()
        self.mon_mer_obstruction = self.make_mon_mer_obstruction_class()
        self.mon_long_obstruction = self.make_mon_long_obstruction_class()
        self.mer_long_obstruction = self.make_mer_long_obstruction_class()
        self.all_obstruction = self.make_all_obstruction_class()
        
        
        #Generate 8 Ptolemy equations to each obstruciton class which is dual to: 
        #     the mapping torus meridean (monodromy),
        #     the meridean of the punctured torus (meridean),
        #     the longitude of the punctured torus (longitude)
        #     combinations of the above
        
        self.trivial_ptolemy = self.replace_all_equations(self.trivial_obstruction, self.ptolemy_equations)
        self.monodromy_ptolemy = self.replace_all_equations(self.monodromy_obstruction, self.ptolemy_equations)
        self.meridean_ptolemy = self.replace_all_equations(self.meridean_obstruction, self.ptolemy_equations)
        self.longitude_ptolemy = self.replace_all_equations(self.longitude_obstruction, self.ptolemy_equations)
        self.mon_mer_ptolemy = self.replace_all_equations(self.mon_mer_obstruction, self.ptolemy_equations)
        self.mon_long_ptolemy = self.replace_all_equations(self.mon_long_obstruction, self.ptolemy_equations)
        self.mer_long_ptolemy = self.replace_all_equations(self.mer_long_obstruction, self.ptolemy_equations)
        self.all_ptolemy = self.replace_all_equations(self.all_obstruction, self.ptolemy_equations)
        
        #Now we work to reduce these equations according to the edge gluings of the triangulations.
        #We start with the easy relabeling of the equations.
        
        self.relabel_dict = self.make_relabel_dict()
        
        self.re_trivial_ptolemy = self.replace_all_equations(self.relabel_dict, self.trivial_ptolemy)
        self.re_monodromy_ptolemy = self.replace_all_equations(self.relabel_dict, self.monodromy_ptolemy)
        self.re_meridean_ptolemy = self.replace_all_equations(self.relabel_dict, self.meridean_ptolemy)
        self.re_longitude_ptolemy = self.replace_all_equations(self.relabel_dict, self.longitude_ptolemy)
        self.re_mon_mer_ptolemy = self.replace_all_equations(self.relabel_dict, self.mon_mer_ptolemy)
        self.re_mon_long_ptolemy = self.replace_all_equations(self.relabel_dict, self.mon_long_ptolemy)
        self.re_mer_long_ptolemy = self.replace_all_equations(self.relabel_dict, self.mer_long_ptolemy)
        self.re_all_ptolemy = self.replace_all_equations(self.relabel_dict, self.all_ptolemy)
        
        self.gluing_dict = self.make_gluing_dict()
        
        self.glued_trivial_ptolemy = self.glue_edges(self.re_trivial_ptolemy)
        self.glued_monodromy_ptolemy = self.glue_edges(self.re_monodromy_ptolemy)
        self.glued_meridean_ptolemy = self.glue_edges(self.re_meridean_ptolemy)
        self.glued_longitude_ptolemy = self.glue_edges(self.re_longitude_ptolemy)
        self.glued_mon_mer_ptolemy = self.glue_edges(self.re_mon_mer_ptolemy)
        self.glued_mon_long_ptolemy = self.glue_edges(self.re_mon_long_ptolemy)
        self.glued_mer_long_ptolemy = self.glue_edges(self.re_mer_long_ptolemy)
        self.glued_all_ptolemy = self.glue_edges(self.re_all_ptolemy)
        
        
        #Export these strings to sage in order to generate a variety and solve.
        
        self.S = self.make_space()
        self.sage_trivial_ptolemy = self.expt_to_sage(self.glued_trivial_ptolemy)
        self.sage_monodromy_ptolemy = self.expt_to_sage(self.glued_monodromy_ptolemy)
        self.sage_meridean_ptolemy = self.expt_to_sage(self.glued_meridean_ptolemy)
        self.sage_longitude_ptolemy = self.expt_to_sage(self.glued_longitude_ptolemy)
        self.sage_mon_mer_ptolemy = self.expt_to_sage(self.glued_mon_mer_ptolemy)
        self.sage_mon_long_ptolemy = self.expt_to_sage(self.glued_mon_long_ptolemy)
        self.sage_mer_long_ptolemy = self.expt_to_sage(self.glued_mer_long_ptolemy)
        self.sage_all_ptolemy = self.expt_to_sage(self.glued_all_ptolemy)
        
        
        
    def make_ptolemy_equations(self, monodromy):
        equations = []
        for w in range(len(monodromy)):
            equations.append(f's_0_{w}*s_3_{w}*c_1001_{w}*c_0110_{w} + s_0_{w}*s_1_{w}*c_1100_{w}*c_0011_{w} - s_0_{w}*s_2_{w}*c_1010_{w}*c_0101_{w}')
        return equations
    
    #Generate obstruction classes. 
    #Notice that I use 'l' instead of '1' for ease of replacement later on.
    
    def list_obstructions(self):
        obstr = {}
        for w in range(len(self.monodromy)):
            obstr[f's_0_{w}'] = 'l'
            obstr[f's_1_{w}'] = 'l'
            obstr[f's_2_{w}'] = 'l'
            obstr[f's_3_{w}'] = 'l'
        return obstr
    
    def make_monodromy_obstruction_class(self):
        end = len(self.monodromy) - 1
        mono_obstr = [f's_1_{end}', 's_3_0', 's_1_0', 's_0_1']
        for w in range(1, end):
            if self.monodromy[w] == 'R':
                mono_obstr.append(f's_1_{w}')
                mono_obstr.append(f's_0_{w+1}')
            elif self.monodromy[w] == 'L':
                mono_obstr.append(f's_1_{w}')
                mono_obstr.append(f's_3_{w+1}')
        
        mono_dict = self.list_obstructions()
        for obstr in mono_obstr:
            mono_dict[obstr] = '(-1)'
        return mono_dict
    
    def make_meridean_obstruction_class(self):
        mer_obstr = ['s_1_0', 's_2_0', 's_0_1', 's_3_1']
        mer_dict = self.list_obstructions()
        for obstr in mer_obstr:
            mer_dict[obstr] = '(-1)'
        return mer_dict
    
    def make_longitude_obstruction_class(self):
        end = len(self.monodromy) - 1
        long_obstr = ['s_0_0', 's_3_0', f's_1_{end}', f's_2_{end}']
        long_dict = self.list_obstructions()
        for obstr in long_obstr:
            long_dict[obstr] = '(-1)'
        return long_dict
    
    def make_mon_mer_obstruction_class(self):
        end = len(self.monodromy) - 1
        mix_obstr = [f's_1_{end}', 's_3_0', 's_2_0', 's_3_1']
        for w in range(1, end):
            if self.monodromy[w] == 'R':
                mix_obstr.append(f's_1_{w}')
                mix_obstr.append(f's_0_{w+1}')
            elif self.monodromy[w] == 'L':
                mix_obstr.append(f's_1_{w}')
                mix_obstr.append(f's_3_{w+1}')
        
        mix_dict = self.list_obstructions()
        for obstr in mix_obstr:
            mix_dict[obstr] = '(-1)'
        return mix_dict
    
    def make_mon_long_obstruction_class(self):
        end = len(self.monodromy) - 1
        mix_obstr = [f's_2_{end}', 's_0_0', 's_1_0', 's_0_1']
        for w in range(1, end):
            if self.monodromy[w] == 'R':
                mix_obstr.append(f's_1_{w}')
                mix_obstr.append(f's_0_{w+1}')
            elif self.monodromy[w] == 'L':
                mix_obstr.append(f's_1_{w}')
                mix_obstr.append(f's_3_{w+1}')
        
        mix_dict = self.list_obstructions()
        for obstr in mix_obstr:
            mix_dict[obstr] = '(-1)'
        return mix_dict
                 
    def make_mer_long_obstruction_class(self):
        end = len(self.monodromy) - 1
        mix_obstr = ['s_0_0', 's_3_0', f's_1_{end}', f's_2_{end}', 's_1_0', 's_2_0', 's_0_1', 's_3_1']
        mix_dict = self.list_obstructions()
        for obstr in mix_obstr:
            mix_dict[obstr] = '(-1)'
        return mix_dict
    
    def make_all_obstruction_class(self):
        end = len(self.monodromy) - 1
        mix_obstr = [f's_2_{end}', 's_0_0', 's_2_0', 's_3_1']
        for w in range(1, end):
            if self.monodromy[w] == 'R':
                mix_obstr.append(f's_1_{w}')
                mix_obstr.append(f's_0_{w+1}')
            elif self.monodromy[w] == 'L':
                mix_obstr.append(f's_1_{w}')
                mix_obstr.append(f's_3_{w+1}')
        mix_dict = self.list_obstructions()
        for obstr in mix_obstr:
            mix_dict[obstr] = '(-1)'
        return mix_dict
    
    #Replace in each equation
                     
    def insert_obstructions(self, obstruction_dict, ptolemy_equation):
        for face, value in obstruction_dict.items():
            ptolemy_equation = ptolemy_equation.replace(face, value)
        return ptolemy_equation
        
    #Replace for each equation in a system
    def replace_all_equations(self, obstruction_dict, ptolemy_equations):
        replaced_equations = []
        for w in range(len(ptolemy_equations)):
            pre_reduced = self.insert_obstructions(obstruction_dict, ptolemy_equations[w])
            reduced = self.simplify_signs(pre_reduced)
            replaced_equations.append(reduced)
        return replaced_equations
    
    #Condense all instances of 1 and -1 into signs
    def simplify_signs(self, ptolemy_equation):
        ptolemy_equation = ptolemy_equation.replace('(-1)*(-1)*', '')
        ptolemy_equation = ptolemy_equation.replace('l*', '')
        ptolemy_equation = ptolemy_equation.replace('+ (-1)*', '- ')
        ptolemy_equation = ptolemy_equation.replace('- (-1)*', '+ ')
        ptolemy_equation = ptolemy_equation.replace('(-1)*', '-')
        return ptolemy_equation
    
    #Next is edge gluing simplifications.
    
    #We can always write all edge-classes in terms of the top diagonal edges in the layered triangulation.
    #i.e. edge c_1001_k.
    
    #The edge identifications will go as follows:
    
    #After an R move, we will get:
    #c_1100_k & c_0011_k <-> c_1100_k+1 & c_0011_k+1
    #c_1010_k & c_0101_k <-> c_0110_k+1
    #c_1001_k            <-> c_1010_k+1 & c_0101_k+1
    
    
    #After an L move, we will get:
    #c_1100_k & c_0011_k <-> -c_0110_k+1
    #c_1010_k & c_0101_k <->  c_1010_k+1 & c_0101_k+1
    #c_1001_k            <->  c_1100_k+1 & c_0011_k+1
    
    #Since the edges of each tetrahedron in this construction are identified as either:
    #horizontal (h),
    #vertical (v),
    #top diagonal (t),
    #bottom diagonal(b),
    #we can substitute these names in for the edges. Moreover, the goal is to simplify the edge classes until only t_k labels remain in these equations.
    
    #Under this new notation, the edge identificaitons appear as:
    
    #After an R move, we will get:
    #h_k <-> h_k+1
    #v_k <-> b_k+1
    #t_k <-> v_k+1
    
    #After an L move, we will get:
    #h_k <-> -b_k+1
    #v_k <->  v_k+1
    #t_k <->  h_k+1
    
    def make_relabel_dict(self):
        relabel_dict = {}
        for w in range(len(self.monodromy)):
            relabel_dict[f'c_1100_{w}'] = f'h_{w}'
            relabel_dict[f'c_0011_{w}'] = f'h_{w}'
            relabel_dict[f'c_1010_{w}'] = f'v_{w}'
            relabel_dict[f'c_0101_{w}'] = f'v_{w}'
            relabel_dict[f'c_1001_{w}'] = f't_{w}'
            relabel_dict[f'c_0110_{w}'] = f'b_{w}'
            
        return relabel_dict
    
    def move_gluing(self, step:int):
        if self.monodromy[step] == 'R':
            return {f'h_{step+1}' : f'h_{step}', f'b_{step+1}' : f'v_{step}', f'v_{step+1}' : f't_{step}'}
        elif self.monodromy[step] == 'L':
            return {f'h_{step+1}' : f't_{step}', f'b_{step+1}' : f'(-1)*h_{step}', f'v_{step+1}' : f'v_{step}'}     
    
    def make_gluing_dict(self):
        end = len(self.monodromy) - 1
        gluing_dict = {'h_0' : f't_{end}', 'v_0' : f'v_{end}', 'b_0' : f'(-1)*h_{end}'}
        for w in range(end):
            gluing_dict = merge(gluing_dict, self.move_gluing(w))
        return gluing_dict
            
    def glue_edges(self, ptolemy_equations):
        for w in range(len(self.monodromy)):
            ptolemy_equations = self.replace_all_equations(self.gluing_dict, ptolemy_equations)
        return Equation(ptolemy_equations).equation
    
    #The following is temporary, uses sage-math to solve the variety equations.
    def make_space(self):
        end = len(self.monodromy)
        S = AffineSpace(end + 1, QQ, 't_')
        return S
    
    def expt_to_sage(self, ptolemy_equations):
        end = len(self.monodromy)
        prod = 't_0'
        for w in range(1, end + 1):
            prod += f'*t_{w}'
        prod += ' - 1'
        ptolemy_equations.append(prod)
        ptolemy_equations.append('t_0 - 1')
        return self.S.subscheme(ptolemy_equations) 
        
    
class Equation():  
    def __init__(self, equation):
        self.equation = equation
        if type(equation) == str:
            self.equation = self.move_minus(self.equation)
        elif type(equation) == list:
            for w in range(len(self.equation)):
                self.equation[w] = self.move_minus(self.equation[w])
        #self.make_powers(self.equation)
        
    def term_start(self, equation, index):
        while index > 0 and equation[index - 1] != ' ':
            index = index - 1
        return index
    
    def term_end(self, equation, index):
        while index < len(equation) - 1 and equation[index + 1] != ' ':
            index = index + 1
        return index
    
    def move_minus(self, equation):
        while '*-' in equation:
            ind = equation.index('*-')
            term_start = self.term_start(equation, ind)
            equation = equation[:term_start] + '-' + equation[term_start: ind+1] + equation[ind+2:]
            
        equation = equation.replace('+ -', '- ')
        equation = equation.replace('- -', '+ ')
        equation = equation.replace('--', '')
        return equation
```
