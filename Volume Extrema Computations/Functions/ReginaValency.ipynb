```python
import snappy
import regina

#Checking the orientation of the edge
def orient(s):
    i = s.index('(')
    if int(s[i+1])<int(s[i+2]):
        return True
    return False
    
#This function prints edge valency data for each edge class in a manifold's triangulation
def edge_val_data(self, label_edges = False, reverse = True):
    M = self
    print(M, M.volume())
    print('Edge valency count:',M.edge_valences(),'\n')
    T = regina.Triangulation3(M._to_string())
    T.makeCanonical()
    e = T.edges()
    
    sort_arry = []
    
    for i in e:
        temp = ''
        l = 0
        b = False
        for k in i:
            l += 1
            if label_edges:
                if b:
                    temp += ', ' + k.str()[0:]
                else:
                    b = True
                    temp += k.str()[0:]
            else:
                if b:
                    if orient(k.str()):
                        temp += ', ' + (' ' + k.str()[0])
                        i = 1
                        while k.str()[i] != ' ':
                            temp += k.str()[i]
                            i += 1
                    else:
                        temp += ', ' + ('-' + k.str()[0])
                        i = 1
                        while k.str()[i] != ' ':
                            temp += k.str()[i]
                            i += 1
                else:
                    b = True
                    if orient(k.str()):
                        temp += (k.str()[0])
                        i = 1
                        while k.str()[i] != ' ':
                            temp += k.str()[i]
                            i += 1
                    else:
                        temp += ('-' + k.str()[0])
                        i = 1
                        while k.str()[i] != ' ':
                            temp += k.str()[i]
                            i += 1
        
        sort_arry.append((l, temp,))
    
    sort_arry.sort(reverse = reverse)
    for i in range(len(sort_arry)):
        # print(f'Edge: {i}, Valency: {sort_arry[i][0]}', f'Gluing: {sort_arry[i][1]}')
        print(f'Valency: {sort_arry[i][0]}\t', f'Gluing: {sort_arry[i][1]}')
        
       
    print('\n' ,'_' * 80, '\n')
    
snappy.Manifold.edge_val_data = edge_val_data
```

```python
#Use Example
snappy.Manifold('b++R'+'L'*4).edge_val_data()
```
