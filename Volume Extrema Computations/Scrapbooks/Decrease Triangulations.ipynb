```python
import snappy
bintoRL = str.maketrans('01', 'RL')

reduced_bool = False
for i in range(10):
    #If a punctured torus bundle is successfully reduced, we do not need to continue, so break.
    if reduced_bool:
        break
        
    #We will generate a random binary integer and convert 0 to R, 1 to L. We will then generate a punctured torus bundle with this.
    word = f'{bin(randint(1,100000000))}'.replace('b', '').translate(bintoRL)
    M = snappy.Manifold('b++'+word)
    
    #Set the initial tetrahedra number which is the number of letters in its RL word
    test = M.num_tetrahedra()

    #Randomize the triangulation and check if the tetrahedra number reduced.
    for k in range(100):
        M.randomize()
        tet_new = M.num_tetrahedra()
        if tet_new < test:
            reduced_bool = True
            print(M, test)
            print(f'Reduced triangulation, new tetrahedra number is {tet_new}')
            break
        
    
    
    
```

```python
import snappy

reduced_bool = False

for M in snappy.OctahedralOrientableCuspedCensus()[:100]:
    test = M.num_tetrahedra()
    if reduced_bool:
        break
    
    #Randomize the triangulation and check if the tetrahedra number reduced.
    for k in range(1000):
        M.randomize()
        tet_new = M.num_tetrahedra()
        if tet_new < test:
            reduced_bool = True
            print(M, test)
            print(f'Reduced triangulation, new tetrahedra number is {tet_new}')
            break

print(reduced_bool)
    
```
