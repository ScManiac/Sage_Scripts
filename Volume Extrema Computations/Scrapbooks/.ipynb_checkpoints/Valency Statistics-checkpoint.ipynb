```python
import snappy

def sdvector(num):
    stddev = []
    for M in snappy.OrientableCuspedCensus(filter=f'tets={num}'):
        val = M.edge_valences()
        stotal = 0
        edges = num
        
        #The mean will always be 6, no point computing
        mean = 6
        
        for k in val.keys():
            stotal += ((k - mean)**2)*val[k]
            
        variance = stotal/edges
        stddev.append((variance**(1/2),M.volume(),))
        
    return stddev

def skewvector(num):
    
    stddev = []
    skew = []
    for M in snappy.OrientableCuspedCensus(filter=f'tets={num}'):
        val = M.edge_valences()
        stotal = 0
        sktotal = 0
        edges = num
        
        #The mean will always be 6, no point computing
        for k in val.keys():
            stotal += ((k - 6)**2)*val[k]
            
        variance = stotal/edges
        sd = variance**(1/2)
        if sd == 0:
            stddev.append((0, M.volume(),))
            skew.append(0)
            break
        
        for k in val.keys():
            sktotal = (((k-6)/sd)**3)*val[k]
        stddev.append((sd, M.volume()))
        skew.append(float(sktotal/edges))
        
        
        
    return (stddev, skew,)
```
