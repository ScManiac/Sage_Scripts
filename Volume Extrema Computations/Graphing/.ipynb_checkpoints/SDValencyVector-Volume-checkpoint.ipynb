```python
import snappy
from sage.plot.colors import red, blue, lime
import pickle

#the data takes a long time to generate, so instead we exported the data to an external file, and import it here
if 'preskew' not in globals() and 'preskew' not in locals():
    with open('preskew.pkl', 'rb') as inp:
        preskew = pickle.load(inp)

#Make the data

def sdvector(num):
    stddev = []
    for M in snappy.OrientableCuspedCensus(filter=f'tets={num}'):
        val = M.edge_valences()
        stotal = 0
        edges = num
        
        #The mean will always be 6, no point computing
        mean = 6
        
        for k in val.keys():
            stotal += ((k - mean)**2)*val[k]
            
        variance = stotal/edges
        stddev.append((variance**(1/2),M.volume(),))
        
    return stddev

def skewvector(num):
    
    stddev = []
    skew = []
    for M in snappy.OrientableCuspedCensus(filter=f'tets={num}'):
        val = M.edge_valences()
        stotal = 0
        sktotal = 0
        edges = num
        
        #The mean will always be 6, no point computing
        for k in val.keys():
            stotal += ((k - 6)**2)*val[k]
            
        variance = stotal/edges
        sd = variance**(1/2)
        if sd == 0:
            stddev.append((0, M.volume(),))
            skew.append(0)
            break
        
        for k in val.keys():
            sktotal = (((k-6)/sd)**3)*val[k]
        stddev.append((sd, M.volume()))
        skew.append(float(sktotal/edges))
        
        
        
    return (stddev, skew,)
        
#Graph function

@interact
def mygraph(numtets = slider(2,9, default = 6),
            skbool = checkbox(False, label = 'Color by Skew'),
    auto_update=False):  

    if skbool:
        data = preskew[numtets]
        # Colors range from lime (left skew) to red (right skew)
        col = [((0,0),'red',)]*len(data[0])

        skmax = max(data[1])
        for i in range(len(col)):
            icol = (data[1][i])/(skmax)
            vcol = lime.blend(red, fraction=icol)
            col[i]=(data[0][i],vcol,)

        plot = False
        for i in col:
            plot += list_plot([(i[0])], color = i[1])
    else:
        data = preskew[numtets][0]
        plot = list_plot(data, color = 'red')
    
    show (plot, scale = 'linear', axes_labels = ['Valency Standard Deviation',f'Volume, Tetrahedra = {numtets}'], frame = True)
```
