```python
import snappy

S = snappy.twister.Surface('S_1_1')

vtet=S.bundle(monodromy=f'(a*B)').volume()/2

print()
print('R times powers of L')
print()
for i in range(10):
    print(f'R*L^{i+1}')
    print(snappy.OrientableCuspedCensus.identify(S.bundle(monodromy=f'a*(B)^{i+1}')))
    print(S.bundle(monodromy=f'a*(B)^{i+1}').volume())
    print()

print('Powers of R*L')
print()
for i in range(10):
    print(f'(R*L)^{i+1}')
    print(snappy.OrientableCuspedCensus.identify(S.bundle(monodromy=f'(a*B)^{i+1}')))
    print(S.bundle(monodromy=f'(a*B)^{i+1}').volume())
    print(f'Max volume with {2*(i+1)} tetrahedra is {vtet*2*(i+1)}')
    print()
    
print()
print('R times powers of R*L')
print()
for i in range(10):
    print(f'R*(R*L)^{i+1}')
    print(snappy.OrientableCuspedCensus.identify(S.bundle(monodromy=f'a*(a*B)^{i+1}')))
    print(S.bundle(monodromy=f'a*(a*B)^{i+1}').volume())
    print(f'Max volume with {2*(i+1)+1} tetrahedra is {vtet*(i+1)*3}')
    print()
    
    
```

```python
sys.path.append('/home/michael/Documents/Math/LaTex/latextablegen/')
import latextablegen
import snappy
headers = ['$# of Tetrahedra$', 'Lowest Volume', 'Lowest Vol. Census Name', '$R \cdot L^{n-1}$', '$Vol(R \cdot L^{n-1})$']
data = []
manifold = 'b++R'
for n in range(2,10):
    manifold += 'L'
    data.append([f'{n}', snappy.OrientableCuspedCensus(filter= f'tets={n}')[0].volume(), snappy.OrientableCuspedCensus(filter= f'tets={n}')[0],  f'$R \cdot L^{{{n-1}}}$', snappy.Manifold(manifold).volume()])

manifold += 'L'
data.append(['10', 'Unknown', 'None',  f'$R \cdot L^{{{9}}}$', snappy.Manifold(manifold).volume()])

tab = latextablegen.LatexTable()

tab.make_file(data = data, headers = headers)
```
