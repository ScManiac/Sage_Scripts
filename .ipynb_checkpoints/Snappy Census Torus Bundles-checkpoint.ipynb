```python
import snappy

S = snappy.twister.Surface('S_1_1')

print('Powers of R*L')
print()
for i in range(10):
    print(f'(R*L)^{i+1}')
    print(snappy.OrientableCuspedCensus.identify(S.bundle(monodromy=f'(a*B)^{i+1}')))
    print(S.bundle(monodromy=f'(a*B)^{i+1}').volume())
    print()
 
print()
print('R times powers of L')
print()
for i in range(10):
    print(f'R*L^{i+1}')
    print(snappy.OrientableCuspedCensus.identify(S.bundle(monodromy=f'a*(B)^{i+1}')))
    print(S.bundle(monodromy=f'a*(B)^{i+1}').volume())
    print()
    
    
# print()
# print('Powers of R times L')
# print()
# for i in range(10):
#     print(f'R{i+1}*L')
#     print(snappy.OrientableCuspedCensus.identify(S.bundle(monodromy=f'(a)^{i+1}*B')))
#     print(S.bundle(monodromy=f'a*(B)^{i+1}').volume())
#     print()
```
