```python
import snappy

S = snappy.twister.Surface('S_1_1')

vtet=S.bundle(monodromy=f'(a*B)').volume()/2

print()
print('R times powers of L')
print()
for i in range(10):
    print(f'R*L^{i+1}')
    print(snappy.OrientableCuspedCensus.identify(S.bundle(monodromy=f'a*(B)^{i+1}')))
    print(S.bundle(monodromy=f'a*(B)^{i+1}').volume())
    print()

print('Powers of R*L')
print()
for i in range(10):
    print(f'(R*L)^{i+1}')
    print(snappy.OrientableCuspedCensus.identify(S.bundle(monodromy=f'(a*B)^{i+1}')))
    print(S.bundle(monodromy=f'(a*B)^{i+1}').volume())
    print(f'Max volume with {2*(i+1)} tetrahedra is {vtet*2*(i+1)}')
    print()
    
print()
print('R times powers of R*L')
print()
for i in range(10):
    print(f'R*(R*L)^{i+1}')
    print(snappy.OrientableCuspedCensus.identify(S.bundle(monodromy=f'a*(a*B)^{i+1}')))
    print(S.bundle(monodromy=f'a*(a*B)^{i+1}').volume())
    print(f'Max volume with {2*(i+1)+1} tetrahedra is {vtet*(i+1)*3}')
    print()
    
    
```

```python
import snappy
import os
from sympy import *

term_size = os.get_terminal_size()

S = snappy.twister.Surface('S_1_1')

for i in range(1,10):
    M = S.bundle(f'a*(B)^{i}')
    print(f'{M} \nVolume = {M.volume()}')
    iterator = 0
    for tet in M.tetrahedra_shapes():
        shape = tet['rect']
        shapep = 1-1/(shape)
        shapepp = 1/(1-shape)
        angle = arg(shape)
        anglep = arg(shapep)
        anglepp = arg(shapepp)
        
        volume = tet['volume']
        print('_' * term_size.columns, f'\nTetrahedra {iterator} \nVolume = {volume} \n\nz = {shape} \nz\' = {shapep} \nz\'\' = {shapepp} \n\n\u03B8 = {angle}\n\u03B8\' = {anglep}\n\u03B8\'\' = {anglepp}')
        print(f'\n\u03C6 = {nsimplify(angle/pi, tolerance=1e-3)}')
        print(f'\u03C6\' = {nsimplify(anglep/pi, tolerance=1e-3)}')
        print(f'\u03C6\'\' = {nsimplify(anglepp/pi, tolerance=1e-3)}')
        iterator += 1
    
    print('=' * term_size.columns, '\n')
```
