```python
import snappy
import regina

def trans(s):
    if s == '01' or s == '10':
        return 'a'
    if s == '02' or s == '20':
        return 'b'
    if s == '03' or s == '30':
        return 'c'
    if s == '12' or s == '21':
        return 'd'
    if s == '13' or s == '31':
        return 'e'
    if s == '23' or s == '32':
        return 'f'

def edge_val_data(self, label_edges = False, reverse = True):
    M = self
    print(M, M.volume(), '\n', M.edge_valences(), '\n')
    T = regina.Triangulation3(M._to_string())
    T.makeCanonical()
    e = T.edges()
    
    sort_arry = []
    
    for i in e:
        temp = ''
        l = 0
        b = False
        for k in i:
            l += 1
            if label_edges:
                if b:
                    if int(k.str()[3]) < int(k.str()[4]):
                        temp += ', ' + (' ' + k.str()[0]) + trans(k.str()[3:5])
                    else:
                        temp += ', ' + ('-' + k.str()[0]) + trans(k.str()[3:5])
                else:
                    b = True
                    if int(k.str()[3]) < int(k.str()[4]):
                        temp += (k.str()[0]) + trans(k.str()[3:5])
                    else:
                        temp += ('-' + k.str()[0]) + trans(k.str()[3:5])
            else:
                if b:
                    if int(k.str()[3]) < int(k.str()[4]):
                        temp += ', ' + (' ' + k.str()[0])
                    else:
                        temp += ', ' + ('-' + k.str()[0])
                else:
                    b = True
                    if int(k.str()[3]) < int(k.str()[4]):
                        temp += (k.str()[0])
                    else:
                        temp += ('-' + k.str()[0])
        
        sort_arry.append((l, temp,))
    
    sort_arry.sort(reverse = reverse)
    for i in range(len(sort_arry)):
        # print(f'Edge: {i}, Valency: {sort_arry[i][0]}', f'Gluing: {sort_arry[i][1]}')
        print(f'Valency: {sort_arry[i][0]}\t', f'Gluing: {sort_arry[i][1]}')
        
       
    print('\n' ,'_' * 80, '\n')
    
snappy.Manifold.edge_val_data = edge_val_data
```

```python
import snappy
import regina

def edge_val_data(self, label_edges = False, reverse = True):
    M = self
    print(M, M.volume())
    print('Edge valency count:',M.edge_valences(),'\n')
    T = regina.Triangulation3(M._to_string())
    T.makeCanonical()
    e = T.edges()
    
    sort_arry = []
    
    for i in e:
        temp = ''
        l = 0
        b = False
        for k in i:
            l += 1
            if label_edges:
                if b:
                    temp += ', ' + k.str()[0:]
                else:
                    b = True
                    temp += k.str()[0:]
            else:
                if b:
                    if int(k.str()[3]) < int(k.str()[4]):
                        temp += ', ' + (' ' + k.str()[0])
                    else:
                        temp += ', ' + ('-' + k.str()[0])
                else:
                    b = True
                    if int(k.str()[3]) < int(k.str()[4]):
                        temp += (k.str()[0])
                    else:
                        temp += ('-' + k.str()[0])
        
        sort_arry.append((l, temp,))
    
    sort_arry.sort(reverse = reverse)
    for i in range(len(sort_arry)):
        # print(f'Edge: {i}, Valency: {sort_arry[i][0]}', f'Gluing: {sort_arry[i][1]}')
        print(f'Valency: {sort_arry[i][0]}\t', f'Gluing: {sort_arry[i][1]}')
        
       
    print('\n' ,'_' * 80, '\n')
    
snappy.Manifold.edge_val_data = edge_val_data
```
