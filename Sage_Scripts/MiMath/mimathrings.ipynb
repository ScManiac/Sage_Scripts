```python
import re
[m.start() for m in re.finditer('test', 'test test test test')]

def find(s, ch):
    return [i.start() for i in re.finditer(ch, s)]

#This is the class of symbolic variables. By calling on its constructor, it will take whatever name is in quotes and assign a variable to it.

class Variable():
    def __init__(self, name):
        self.name = str(name)
        globals()[name] = self
        return None

    def __repr__(self):
        return self.name
    
    def __str__(self):
        return self.name

#This is the class of ring elements. It defines how we can do algebraic opperations on ring expressions.
    
class RingElement(Variable):
    def __init__(self, name):
        self.name = str(name).replace(' ','')
        self._remove_sub()
        self._merge_sub()
        self.terms = self.terms()
        
        
    #Takes a '-' symbol and determines if it is a negative or subtraction.    
    def _is_sub(self, index):
        string = self.name
        preceding = string[index - 1]
        
        if string[index] != '-':
            raise ValueError(f"{string} does not have a '-' at index {index}") 
            
        #First character is '-'    
        if index == 0:
            return False
        
        #Immediately following a parenthesis
        elif preceding == '(':
            return False
        
        #Immediately following exponentiation
        elif preceding == '^':
            return False
        
        #Immediately following multiplication
        elif preceding == '*':
            return False
        
        #Immediately following division
        elif preceding == '/':
            return False
        
        #Immediately following addition
        elif preceding == '+':
            return False
        
        #Immediately following subtraction
        elif preceding == '-':
            return False
        
        else:
            return True
     
    #Find all instances of subtraction within the expression
    def _find_sub(self):
        return [minus for minus in find(self.name, '-') if self._is_sub(minus)]
    
    #Replace all instances of subtraction with '+-'
    def _remove_sub(self):
        for sub in reversed(self._find_sub()):
            self.name = self.name[:sub] + '+-' + self.name[sub+1:]
        return None
    
    #Remove all instances of '--'
    def _merge_sub(self):
        dsub = find(self.name, '--')
        for sub in reversed(dsub):
            self.name = self.name[:sub] + '' + self.name[sub+2:]
        return None
    
    #List of terms in the expression    
    def terms(self):
        return [term for term in self.name.split('+')]
    
    def factors(self):
        return [term.split('*') for term in self.terms]
        
        
    def __repr__(self):
        return self.name
    
    def __str__(self):
        return self.name
    
    

                

class Ring():
    def __init__(self, *variables: str):
        for var in variables:
            RingElement(var)
        return None
            
    def __call__(self, ring_element):
        return None
            
        

class PolynomialRing():
    def __init__(self, base_ring: Ring):
        return None
```
