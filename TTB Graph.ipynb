```python
import snappy 
import numpy as np

def gen_values(rng):        
    values = []

    for i in range(2,rng+1):
        values.append(i)

    return values

@parallel
def compute_points(num):
    mfld = snappy.Manifold('b++R'+'L'*(num-1))
    out_arry = [0,*np.zeros(4)]
    volume = mfld.volume()
    shape_vol = mfld.tetrahedra_shapes('volume')
    out_arry[0] = num
    out_arry[1] = volume
    out_arry[2] = min(shape_vol)
    out_arry[3] = max(shape_vol)
    out_arry[4] = volume/(num)

    return out_arry

def gen_points(rng):
    pre_out = list(zip(*list(zip(*compute_points(gen_values(rng))))[1]))
    ind = pre_out.pop(0)
    return [list(zip(ind, pre_out[0])), list(zip(ind, pre_out[1])), list(zip(ind, pre_out[2])), list(zip(ind, pre_out[3]))]

```
